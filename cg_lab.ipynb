# Computer Graphics Lab

Line and circle drawing algorithms:
1. DDA Line Algorithm
2. Bresenham Line Algorithm
3. Bresenham Circle Algorithm
4. Midpoint Circle Algorithm

## 1. DDA Line Algorithm

```python
import matplotlib.pyplot as plt

def dda_line(x1, y1, x2, y2):
    dx, dy = x2 - x1, y2 - y1
    steps = max(abs(dx), abs(dy))
    x_inc, y_inc = dx / steps, dy / steps
    x, y = x1, y1
    for _ in range(steps + 1):
        print("Pixel", round(x), round(y))
        plt.plot(round(x), round(y), 'gs')
        x += x_inc
        y += y_inc
    plt.plot([x1, x2], [y1, y2], color="red")
    plt.title("DDA Line")
    plt.grid(True)
    plt.axis('equal')
    plt.show()

dda_line(2, 2, 15, 8)
```

## 2. Bresenham Line Algorithm

```python
def bresenham_line(x1, y1, x2, y2):
    dx, dy = x2 - x1, y2 - y1
    p = 2 * dy - dx
    x, y = x1, y1
    while x <= x2:
        print("Pixel", x, y)
        plt.plot(x, y, 'gs')
        if p < 0:
            p = p + 2 * dy
        else:
            y = y + 1
            p = p + 2 * dy - 2 * dx
        x = x + 1
    plt.plot([x1, x2], [y1, y2], color="red")
    plt.title("Bresenham Line")
    plt.grid(True)
    plt.axis('equal')
    plt.show()

bresenham_line(2, 2, 15, 8)
```

## 3. Bresenham Circle Algorithm

```python
def bresenham_circle(xc, yc, r):
    x, y = 0, r
    d = 3 - 2 * r
    while x <= y:
        for px, py in [(xc+x, yc+y), (xc-x, yc+y), (xc+x, yc-y), (xc-x, yc-y),
                       (xc+y, yc+x), (xc-y, yc+x), (xc+y, yc-x), (xc-y, yc-x)]:
            print("Pixel", px, py)
            plt.plot(px, py, 'gs')
        if d < 0:
            d += 4 * x + 6
        else:
            d += 4 * (x - y) + 10
            y -= 1
        x += 1
    circle = plt.Circle((xc, yc), r, color="red", fill=False)
    plt.gca().add_patch(circle)
    plt.title("Bresenham Circle")
    plt.grid(True)
    plt.axis('equal')
    plt.show()

bresenham_circle(0, 0, 10)
```

## 4. Midpoint Circle Algorithm

```python
def midpoint_circle(xc, yc, r):
    x, y = 0, r
    p = 1 - r
    while x <= y:
        for px, py in [(xc+x, yc+y), (xc-x, yc+y), (xc+x, yc-y), (xc-x, yc-y),
                       (xc+y, yc+x), (xc-y, yc+x), (xc+y, yc-x), (xc-y, yc-x)]:
            print("Pixel", px, py)
            plt.plot(px, py, 'gs')
        if p < 0:
            p += 2 * x + 3
        else:
            p += 2 * (x - y) + 5
            y -= 1
        x += 1
    circle = plt.Circle((xc, yc), r, color="red", fill=False)
    plt.gca().add_patch(circle)
    plt.title("Midpoint Circle")
    plt.grid(True)
    plt.axis('equal')
    plt.show()

midpoint_circle(0, 0, 10)
```
